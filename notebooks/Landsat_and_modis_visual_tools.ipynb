# Automated tool for MODIS and Landsat land surface temperature compositing, visualisation, and export

The following function automates the following process for the MODIS and Landsat 8/9 land surface product
- Filters observations by time from both MODIS and Landsat-8/9
- Filters observations by region of interest geometry
- Processes by masking invalid pixels because of clouds and other quality flags
- Performs temporal compositing. This could be weekly, monthly, or any user-specified date
- Computes the number of days the temperature exceeds a specific threshold. Please note that this outcome will also be affected by the number of valid observations
- Visualises the computed time series. This could be either composite at a specific time or within an entire specific period
- Provides a tool to either select or mass-export data to Google Drive


To properly use the tool, first import the wrapper function indicated in the subsequent cell. The imported functions will take the following parameters

-`roi`: which is the region of interest definition geometry or file. If its geometry it should be any polygon definition, see example or an absolute file path to .shp, .geojson, .kml files

-`start_date`: the start date for searching the image. This should be in 'YYYY-MM-DD' format

-`end_date`: the end date for searching the image. This should be in 'YYYY-MM-DD' format

-`aggregation_time`: the temporal aggregation scale. This could be "weekly", "8-day", "monthly" or any specific number of days, like 15

-`aggregation_stat`: the aggregation statistics for temporal aggregation. This could be mean, median, max, min or stdev

-`temp_threshold`: the threshold to count the number of days the temperature exceeds a specific threshold 

-`process_type`: for MODIS to apply whether strict or relaxed masks for quality control

-`export_prefix`: The file name prefix while exporting the images to Google Drive

-`sensor_id`: The MODIS sensor ID. This could be either 'aqua' or 'terra'


## Usage

Import the wraper functions

```python
import ee
from visual_tools.landsat_processor import automate_landsat
from visual_tools.modis_processor import automate_modis
```

#### Example to define earth engine geometry file
This is only important incase there is no predefined area of interest file localy

```python
roi_span = ee.Geometry.Polygon([
    [[16.18, 48.12], [16.59, 48.12], [16.59, 48.34], [16.18, 48.34], [16.18, 48.12]]
])
```

or use file directly from local folder 

```python
# roi_span = "../roi_span.shp"
```

#### Landsat

```python
_, _, _, _, = automate_landsat(
    roi=roi_span,
    start_date='2021-01-01',
    end_date='2021-12-30',
    aggregation_time='monthly',
    aggregation_stat="mean",
    temp_threshold=30.0)
```

#### MODIS

```python
_, _, _, _ = automate_modis(
    roi=roi_span,
    start_date='2021-01-01',
    end_date='2021-12-30',
    aggregation_time='monthly',
    aggregation_stat="max",
    temp_threshold=30.0,
    process_type='relaxed',
    export_prefix='MODIS_LST',
    sensor_id = "aqua"
)
```
